```python
# We have 122 categorical columns, which we should one-hot encode
ohc_cat = pd.get_dummies(df[cat_cols], drop_first=True)
ohc_cat.shape
```

```python
#how many new columns were created
plus_cols=ohc_cat.shape[1] - df[cat_cols].shape[1]
plus_cols
```

```python
store_df_2 = ohc_cat.columns[ohc_cat.columns.str.contains('store', case=False)]
store_df_2
```

```python
#lets see the correlation of these variables with the target variable, store type
import seaborn as sns
import matplotlib.pyplot as plt

store_types = ohc_cat[store_df_2]
```

```python
store_types.head()
```

```python
ohc_cat=ohc_cat.astype(int)
```

```python
store_df=store_types[['store_type_Hypermarket', 'store_type_Neighborhood Store',
       'store_type_Supermarket', 'store_type_Wholesale Store']]
```

```python
corr_matrix = pd.DataFrame(
    index=store_df.columns,
    columns=ohc_cat.columns,
    dtype=float
)

for col1 in store_df.columns:
    for col2 in ohc_cat.columns:
        corr_matrix.loc[col1, col2] = store_df[col1].corr(ohc_cat[col2])

pd.DataFrame(corr_matrix).T
```
