```python
#counting how many columns would be created if one hot encoding is applied to the categorical columns in the dataframe
#select the object type columns in the dataframe
mask=df_supmark.dtypes==object
categorical_cols=df_supmark.columns[mask]
```

```python
num_ohc_cols=(df_supmark[categorical_cols].apply(lambda x: x.nunique()).sort_values(ascending=False))
```

```python
#no need to encode those, which have only 1 unique value, as they will not add any information to the model.
small_num_ohc_cols=num_ohc_cols.loc[num_ohc_cols>1]
small_num_ohc_cols-=1 #ohc will create n-1 columns for each categorical column

#sum the number of these columns
small_num_ohc_cols.sum()
```
