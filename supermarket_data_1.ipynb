```python
nuniq = df_supmark.select_dtypes(include='object').nunique()
```

```python
cols_to_encode = nuniq[(nuniq > 1) & (nuniq < 10)].index.tolist()
print(len(cols_to_encode), "oszlop lesz kódolva")
```

```python
print("Várható új dummy oszlopok:", (nuniq[cols_to_encode] - 1).sum())
```

```python
df_supmark_encoded = pd.get_dummies(df_supmark, columns=cols_to_encode, drop_first=True)
print(df_supmark.shape, "->", df_supmark_encoded.shape)
print(df_supmark_encoded.columns[-10:].tolist())
```
