```python
# the remaining object type columns will be excluded from the dataaframe as they may conatin string values
# that are not suitable for one hot encoding. These columns will be dropped from the dataframe.
X = df_supmark_encoded.select_dtypes(exclude='object')
```

```python
# due to large number of rows, we should reduce the number aof variables with PCA.
X.isnull().sum().sort_values(ascending=False).head(20)
```

```python
X.shape
```

```python
# from 120000 records each column, several have more than 1000 missing values, so wee have to handle those
# lets drop those rows with missing values, as they are only a small fraction of the total data:]
X2=X.dropna()
```

```python
X2.isnull().sum().sort_values(ascending=False)
```

```python
X3 = X2.select_dtypes(include=['number'])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X3)
```

```python
from sklearn.decomposition import PCA

pca = PCA()

X_pca = pca.fit_transform(X_scaled)
```

```python
import numpy as np
import matplotlib.pyplot as plt

explained_variance = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker='.'
)

plt.axhline(y=0.90, linestyle='--', label='90%')
plt.axhline(y=0.95, linestyle='--', label='95%')

plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA - Cumulative Explained Variance')

plt.legend()
plt.grid()
plt.show()
```

```python
pca = PCA(n_components=0.90)

X_pca = pca.fit_transform(X_scaled)

print("Eredeti változók:", X3.shape[1])
print("Főkomponensek:", X_pca.shape[1])
print("Megőrzött variancia:",
      pca.explained_variance_ratio_.sum())
```

```python
# putting the PCA results into a dataframe
pca_columns = [
    f'PC{i+1}' for i in range(X_pca.shape[1])
]

data_pca = pd.DataFrame(
    X_pca,
    columns=pca_columns,
    index=X3.index
)
```

```python

loadings = pd.DataFrame(
    pca.components_.T,
    columns=pca_columns,
    index=X3.columns
)
loadings
```

```python
# showing the top 30 features that contribute to the first component
print(
    loadings['PC1']
    .abs()
    .sort_values(ascending=False)
    .head(30)
)
```

```python
# finding out the most freqent variables
from collections import Counter

top_variables = []

for pc in loadings.columns:
    top_15 = loadings[pc].abs().nlargest(15).index.tolist()
    top_variables.extend(top_15)
```

```python
variable_counts = Counter(top_variables)

most_common = pd.DataFrame(
    variable_counts.most_common(),
    columns=['Variable', 'Frequency']
)

print(most_common.head(30))
```
